# Exploratory Data Analysis (EDA)
Step-by-step EDA guide for your dataset.

### Step 1: Import Required Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Visualization configuration
sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (10, 6)
print("Libraries loaded successfully!")
```

### Step 2: Load Your Dataset
Put your csv filename below:

```python
# Put your dataset filename here (e.g. 'data.csv')
df = pd.read_csv("your_dataset.csv")

# Display first 5 rows
df.head()
```

### Step 3: Dataset Shape & Overview

```python
print(f"Total Rows: {df.shape[0]}, Total Columns: {df.shape[1]}")
print("\n--- Info ---")
df.info()
```

### Step 4: Summary Statistics

```python
# Numerical columns summary
df.describe()
```

```python
# Categorical columns summary (if any)
df.describe(include=["O"])
```

### Step 5: Check Missing & Duplicate Values

```python
missing = df.isnull().sum()
missing_percent = (df.isnull().sum() / len(df)) * 100
missing_df = pd.DataFrame({"Missing Count": missing, "Percentage (%)": missing_percent})
print("Missing Values:")
print(missing_df[missing_df["Missing Count"] > 0])

print(f"\nDuplicate Rows Count: {df.duplicated().sum()}")
```

### Step 6: Identify Column Types

```python
num_cols = df.select_dtypes(include=np.number).columns.tolist()
cat_cols = df.select_dtypes(include=["object", "category"]).columns.tolist()

print(f"Numerical Columns ({len(num_cols)}):", num_cols)
print(f"Categorical Columns ({len(cat_cols)}):", cat_cols)
```

### Step 7: Numerical Distribution Plots

```python
if num_cols:
    df[num_cols].hist(figsize=(14, 8), bins=20)
    plt.suptitle("Distributions of Numerical Features", fontsize=14)
    plt.tight_layout()
    plt.show()
```

### Step 8: Categorical Columns Value Counts & Bar Charts

```python
for col in cat_cols:
    if df[col].nunique() <= 15:
        plt.figure(figsize=(8, 4))
        sns.countplot(data=df, x=col, order=df[col].value_counts().index)
        plt.title(f"Distribution of {col}")
        plt.xticks(rotation=45)
        plt.show()
```

### Step 9: Correlation Matrix & Heatmap

```python
if len(num_cols) > 1:
    plt.figure(figsize=(10, 8))
    sns.heatmap(df[num_cols].corr(), annot=True, fmt=".2f", cmap="coolwarm", square=True)
    plt.title("Correlation Heatmap")
    plt.show()
```

### Step 10: Outlier Detection (Boxplots)

```python
if num_cols:
    plt.figure(figsize=(14, 6))
    sns.boxplot(data=df[num_cols])
    plt.title("Boxplot for Outlier Detection")
    plt.xticks(rotation=45)
    plt.show()
```
